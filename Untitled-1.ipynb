## Analogi: bingkai foto dan lembar grafik

Bayangkan Anda menempel beberapa grafik di satu papan.

- **`fig` (figure)** = **papan atau kanvas besarnya**. Ini wadah untuk seluruh gambar.
- **`ax` (axes)** = **satu kotak grafik di atas papan itu**. Sumbu x, sumbu y, garis, dan judul semuanya ada di dalam sebuah `ax`.
- **subplot** = istilah untuk **satu grafik kecil di dalam papan yang berisi beberapa grafik**. Jadi satu `ax` sama dengan satu subplot.

Sebuah `fig` bisa berisi satu `ax` atau banyak `ax`:

```
fig (papan besar)
┌─────────────────────────────────┐
│  ax[0]            ax[1]         │
│ ┌──────────┐    ┌──────────┐    │
│ │ grafik 1 │    │ grafik 2 │    │
│ └──────────┘    └──────────┘    │
└─────────────────────────────────┘
```

Grafik yang Anda gambar tidak menempel langsung di papan, tapi di dalam kotak (`ax`).

## Membaca sintaks barisnya

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
```

### `plt.subplots(...)`
Fungsi ini artinya "siapkan papan dan kotak-kotak grafiknya". Perhatikan huruf **s** di akhir. Ini berbeda dengan `plt.subplot` (tanpa s), yang cara pakainya lebih kuno. Pakai yang berakhiran **s**.

### Tiga argumen di dalamnya

| Argumen | Artinya | Nilai di contoh |
|---|---|---|
| Argumen 1 | jumlah **baris** kotak | `1` |
| Argumen 2 | jumlah **kolom** kotak | `2` |
| `figsize=(10, 4)` | ukuran papan (lebar, tinggi) dalam inci | lebar 10, tinggi 4 |

Jadi `(1, 2)` artinya 1 baris dan 2 kolom, atau dua kotak berdampingan ke kanan. Kalau `(2, 1)`, dua kotak bertumpuk ke bawah. Kalau `(2, 2)`, ada empat kotak.

### `fig, ax = ...`
Fungsi `plt.subplots` mengembalikan **dua benda sekaligus**: papan dan kotak-kotaknya. Python membolehkan kita menampung dua nilai itu ke dua nama variabel:

- `fig` menampung papannya.
- `ax` menampung kotak-kotaknya.

Anda bisa memberi nama lain (misalnya `papan, kotak = plt.subplots(...)`), tapi `fig, ax` adalah kebiasaan yang dipakai semua orang, jadi ikuti saja.

## Isi `ax` tergantung jumlah kotak

| Yang dipanggil | Isi `ax` | Cara mengaksesnya |
|---|---|---|
| `plt.subplots()` atau `(1, 1)` | **satu** kotak | langsung `ax` |
| `plt.subplots(1, 2)` | daftar 2 kotak | `ax[0]` dan `ax[1]` |
| `plt.subplots(2, 2)` | tabel 2x2 kotak | `ax[0, 0]`, `ax[0, 1]`, `ax[1, 0]`, `ax[1, 1]` |

Penomoran dimulai dari **0**, jadi `ax[0]` adalah kotak **pertama** (kiri) dan `ax[1]` adalah kotak **kedua** (kanan). Pada 2x2, `ax[baris, kolom]`, dan `ax[0, 1]` berarti baris pertama, kolom kedua.

## Cara menggambar di sebuah kotak

Setelah punya `ax`, Anda menggambar dengan memanggil method milik kotak itu:

```python
ax[0].plot(x, y)              # gambar garis di kotak kiri
ax[0].set_title('Judul kiri') # judul kotak kiri
ax[0].set_xlabel('Sumbu x')   # label sumbu x kotak kiri
ax[0].set_ylabel('Sumbu y')   # label sumbu y kotak kiri
```

Perhatikan bahwa kalau memakai `ax`, namanya berubah sedikit:

| Gaya `plt` (satu grafik) | Gaya `ax` (banyak grafik) |
|---|---|
| `plt.title(...)` | `ax.set_title(...)` |
| `plt.xlabel(...)` | `ax.set_xlabel(...)` |
| `plt.ylabel(...)` | `ax.set_ylabel(...)` |
| `plt.plot(...)` | `ax.plot(...)` |
| `plt.legend()` | `ax.legend()` |
| `plt.axhline(...)` | `ax.axhline(...)` |

Kenapa ada dua gaya? Saat memanggil `plt.plot(...)` tanpa membuat `fig` dan `ax`, matplotlib diam-diam membuatkan satu papan dengan satu kotak untuk Anda. Itu sebabnya kode PR curve tadi bisa jalan tanpa `fig, ax`. Tapi kalau ingin **banyak grafik berdampingan**, Anda harus menentukan sendiri mau menggambar di kotak yang mana, dan di situlah `ax` diperlukan.

## Contoh kecil yang bisa langsung dicoba

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(10, 4))

ax[0].set_title('Kotak kiri')

ax[1].plot([1, 2, 3], [3, 2, 1])
ax[1].set_title('Kotak kanan')

```

Hasilnya dua grafik berdampingan dengan judul berbeda. Coba ubah `(1, 2)` menjadi `(2, 1)` dan lihat apa yang berubah.

## Dua tambahan yang sering berguna

- **`plt.tight_layout()`** (sebelum `plt.show()`) merapikan jarak antar kotak supaya judul dan label tidak saling menimpa.
- **`fig.suptitle('Judul besar')`** memberi judul untuk seluruh papan, bukan satu kotak.

## Ringkasan satu kalimat

`fig` adalah papan besarnya, `ax` adalah kotak grafik di atas papan, dan `subplots(baris, kolom)` menyiapkan keduanya. Untuk menggambar di kotak tertentu, panggil `ax[nomor]` lalu method-nya (`plot`, `set_title`, dan seterusnya).

Coba jalankan contoh kecil tadi dulu. Kalau sudah, ubah menjadi `plt.subplots(2, 2)` dan isi keempat kotaknya (ingat aksesnya `ax[0, 0]` dan seterusnya). Kalau ada bagian yang masih membingungkan, tunjukkan bagian yang mana.
